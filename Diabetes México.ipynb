Limpieza del conjunto de datos construido a partir de informacion pública de la **Encuesta Nacional de Salud y Nutrición (ENSANUT) 2024**, publicada por el Instituto Nacional de Salud Pública (INSP) de México.\
url: https://ensanut.insp.mx/encuestas/ensanutcontinua2024/descargas.php

```python
import numpy as np
import pandas as pd
```

```python
db = pd.read_csv('Diabetes_Mexico.csv')
```

```python
col_con_comas = db[['Peso','Estatura','ac_urico','creat','insulina','albu','ponde_venosa','ponde_hemo','col_hdl','col_ldl','colest','glu_suero','trig','hb1ac','estrato','']]
```

```python
# Cambiamos las columnas con comas por puntos decimales
for col in col_con_comas:
    if db[col].dtype == 'O': #Si es de tipo Object (texto)
        db[col] = db[col].str.replace(',','.').astype(float)
```

```python
# Eliminamos los valores nulos en las columnas 'sexo', 'Peso' y 'Estatura'
for col in db[['edad','Peso','Estatura']]:
    db[col] = db.groupby('sexo')[col].transform(lambda x:x.fillna(x.mean()))
```

```python
# Calculamos la columna imc
db['imc'] = db['Peso']/(db['Estatura']/100)**2
db
```

```python
# Clasificamos el imc por categorías
def clasif(imc):
    if imc<18.5:
        return 'Bajo Peso'
    elif imc<24.9:
        return 'Normal'
    elif imc<29.9:
        return 'Sobrepeso'
    else:
        return 'Obesidad'
```

```python
db['imc'] = db['imc'].apply(clasif)
```
